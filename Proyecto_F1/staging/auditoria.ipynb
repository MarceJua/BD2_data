```python
import pandas as pd

bios = pd.read_csv('../raw/bios.csv')
results = pd.read_csv('../raw/results.csv')
bios_locs = pd.read_csv('../raw/bios_locs.csv')
noc_galli = pd.read_csv('../raw/noc_regions.csv')
poblacion = pd.read_csv('../raw/populations.csv')

athlete_events = pd.read_csv('../raw/kaggle_athlete_events.csv')
noc_kaggle = pd.read_csv('../raw/kaggle_noc_regions.csv')

summer = pd.read_csv('../raw/kaggle_olympics_dataset.csv')

datacamp = pd.read_csv('../raw/datacamp_athlete_events.csv.gz', compression='gzip')
```

```python
print("Medal - results.csv:", results['Medal'].unique())
print("Medal - athlete_events.csv:", athlete_events['Medal'].unique())
print("Medal - summer:", summer['Medal'].unique())
print("Medal - datacamp:", datacamp['medal'].unique())
```

```python
set_galli = set(noc_galli['NOC'].dropna().unique())
set_kaggle = set(noc_kaggle['NOC'].dropna().unique())

print("¿Son exactamente el mismo catálogo?", set_galli == set_kaggle)
print("Códigos solo en Galli:", set_galli - set_kaggle)
print("Códigos solo en Kaggle:", set_kaggle - set_galli)
```
