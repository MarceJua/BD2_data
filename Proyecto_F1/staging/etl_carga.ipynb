```python
import pandas as pd
from sqlalchemy import create_engine

servidor = "localhost"
base_datos = "OlimpiadasDB"

cadena_conexion = f"mssql+pyodbc://@{servidor}/{base_datos}?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes"
motor = create_engine(cadena_conexion)

print("Conexión creada")
```

```python
noc_regions = pd.read_csv('../raw/noc_regions.csv')
poblacion_raw = pd.read_csv('../raw/populations.csv')

noc_regions['region_norm'] = noc_regions['region'].str.strip().str.lower()
poblacion_raw['country_norm'] = poblacion_raw['Country Name'].str.strip().str.lower()

cruce = noc_regions.merge(
    poblacion_raw[['Country Name', 'Country Code', 'country_norm']],
    left_on='region_norm', right_on='country_norm', how='left'
)

equivalencia = cruce[['NOC', 'region', 'Country Code', 'Country Name']].copy()
equivalencia.columns = ['codigo_noc', 'nombre_noc', 'codigo_iso3', 'nombre_iso3']
equivalencia['revisado'] = equivalencia['codigo_iso3'].notna().map({True: 'SI', False: 'NO'})

print(equivalencia['revisado'].value_counts())
```

```python
equivalencia.to_sql('stg_equivalencia_paises', con=motor, if_exists='append', index=False, chunksize=1000)
print("Cargado a stg_equivalencia_paises")
```

```python
print(equivalencia[equivalencia['nombre_noc'].isnull()])
```

```python
mapa_manual = {
    'TUV': 'Tuvalu',
    'ROT': 'Refugee Olympic Team',
    'UNK': 'Unknown'
}

equivalencia.loc[equivalencia['codigo_noc'].isin(mapa_manual.keys()), 'nombre_noc'] = \
    equivalencia['codigo_noc'].map(mapa_manual)

# Por si en el futuro apareciera algún otro código sin nombre que no esté en el mapa,
# este fillna sigue como respaldo (no debería activarse ahora)
equivalencia['nombre_noc'] = equivalencia['nombre_noc'].fillna(equivalencia['codigo_noc'])

print(equivalencia[equivalencia['codigo_noc'].isin(['TUV','ROT','UNK'])])
```

```python
equivalencia.to_sql('stg_equivalencia_paises', con=motor, if_exists='append', index=False, chunksize=1000)
print("Cargado a stg_equivalencia_paises")
```

```python
columnas_id = ['Country Name', 'Country Code']
columnas_anio = [c for c in poblacion_raw.columns if c not in columnas_id + ['country_norm']]

poblacion_larga = poblacion_raw.melt(
    id_vars=columnas_id,
    value_vars=columnas_anio,
    var_name='anio',
    value_name='poblacion'
)

poblacion_larga = poblacion_larga.dropna(subset=['poblacion'])
poblacion_larga.columns = ['country_name', 'country_code', 'anio', 'poblacion']

print(poblacion_larga.shape)
poblacion_larga.head()
```

```python
poblacion_larga.to_sql('stg_poblacion', con=motor, if_exists='append', index=False, chunksize=1000)
print("Cargado a stg_poblacion")
```

```python
import pandas as pd
from sqlalchemy import create_engine

servidor = "localhost"
base_datos = "OlimpiadasDB"
cadena_conexion = f"mssql+pyodbc://@{servidor}/{base_datos}?driver=ODBC+Driver+17+for+SQL+Server&trusted_connection=yes"
motor = create_engine(cadena_conexion)

athlete_events = pd.read_csv('../raw/kaggle_athlete_events.csv')
print(athlete_events.shape)
```

```python
conteo = athlete_events.dropna(subset=['Medal']).groupby(['Event','Games','Medal']).size().reset_index(name='conteo')
eventos_equipo = conteo[conteo['conteo'] > 1][['Event']].drop_duplicates()
print(len(eventos_equipo))
eventos_equipo.head(10)
```

```python
conteo = athlete_events.dropna(subset=['Medal']).groupby(['Event','Games','Medal']).size().reset_index(name='conteo')
eventos_equipo = conteo[conteo['conteo'] > 3][['Event']].drop_duplicates()
print(len(eventos_equipo))
eventos_equipo.head(10)
```

```python
eventos_equipo.to_sql('stg_eventos_equipo', con=motor, if_exists='append', index=False, chunksize=1000)
print("Cargado a stg_eventos_equipo")
```

```python
import re

atletas_existentes = pd.read_sql("SELECT id_atleta, nombre, id_origen_bios FROM ATLETA", motor)
bios_locs_noc = pd.read_sql("SELECT athlete_id, NOC FROM stg_bios_locs", motor)
atletas_existentes = atletas_existentes.merge(
    bios_locs_noc, left_on='id_origen_bios', right_on='athlete_id', how='left'
)

def normalizar(nombre):
    if pd.isna(nombre):
        return ''
    nombre = str(nombre).lower().strip()
    nombre = re.sub(r'[^a-z ]', '', nombre)
    return nombre

atletas_existentes['clave'] = atletas_existentes['nombre'].apply(normalizar) + '|' + atletas_existentes['NOC'].fillna('').str.lower()

kaggle_atletas = athlete_events[['ID','Name','Sex','Height','Weight','NOC']].drop_duplicates(subset=['ID'])
kaggle_atletas['clave'] = kaggle_atletas['Name'].apply(normalizar) + '|' + kaggle_atletas['NOC'].fillna('').str.lower()

emparejados = kaggle_atletas.merge(atletas_existentes[['id_atleta','clave']], on='clave', how='inner')
nuevos = kaggle_atletas[~kaggle_atletas['clave'].isin(atletas_existentes['clave'])]

print("Emparejados (ya existían en Galli):", len(emparejados))
print("Nuevos (solo están en Kaggle):", len(nuevos))
```

```python
atletas_existentes = pd.read_sql("""
    SELECT a.id_atleta, a.nombre AS nombre_atleta, a.id_origen_bios, p.codigo_noc
    FROM ATLETA a
    INNER JOIN stg_bios_locs bl ON bl.athlete_id = a.id_origen_bios
    LEFT JOIN PAIS p ON p.nombre = bl.NOC
""", motor)

atletas_existentes['clave'] = atletas_existentes['nombre_atleta'].apply(normalizar) + '|' + atletas_existentes['codigo_noc'].fillna('').str.lower()

kaggle_atletas = athlete_events[['ID','Name','Sex','Height','Weight','NOC']].drop_duplicates(subset=['ID'])
kaggle_atletas['clave'] = kaggle_atletas['Name'].apply(normalizar) + '|' + kaggle_atletas['NOC'].fillna('').str.lower()

emparejados = kaggle_atletas.merge(atletas_existentes[['id_atleta','clave']], on='clave', how='inner')
nuevos = kaggle_atletas[~kaggle_atletas['clave'].isin(atletas_existentes['clave'])]

print("Emparejados (ya existían en Galli):", len(emparejados))
print("Nuevos (solo están en Kaggle):", len(nuevos))
```

```python
emparejados_unicos = emparejados.drop_duplicates(subset=['ID'], keep='first')[['id_atleta','ID']]
emparejados_unicos.columns = ['id_atleta','id_kaggle']
print(len(emparejados_unicos))
```

```python
emparejados_unicos.to_sql('stg_atletas_emparejados', con=motor, if_exists='append', index=False, chunksize=1000)
print("Cargado a stg_atletas_emparejados")
```

```python
nuevos_para_insertar = nuevos[['Name','Sex','Height','Weight','ID']].copy()
nuevos_para_insertar.columns = ['nombre','genero','altura','peso','id_origen_kaggle']
nuevos_para_insertar['id_origen_bios'] = None

print(len(nuevos_para_insertar))
```

```python
nuevos_para_insertar.to_sql('stg_atletas_nuevos', con=motor, if_exists='append', index=False, chunksize=1000)
print("Cargado a stg_atletas_nuevos")
```
